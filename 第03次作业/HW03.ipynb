# 第一题


```python
import matplotlib.pyplot as plt

from sklearn.datasets import make_circles
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score
from sklearn.metrics import normalized_mutual_info_score
from sklearn.metrics import adjusted_rand_score


# 生成数据
x, y = make_circles(
    n_samples=400,
    noise=0.05,
    factor=0.4,
    random_state=42
)

# 建立 K-Means 模型
kmeans = KMeans(
    n_clusters=2,
    random_state=42
)

# 进行聚类
kmeans.fit(x)

# 得到聚类结果
y_pred = kmeans.labels_

# 得到聚类中心
centers = kmeans.cluster_centers_


# 可视化
plt.figure(figsize=(12, 5))

# 原始数据
plt.subplot(1, 2, 1)

plt.scatter(
    x[:, 0],
    x[:, 1],
    c=y
)

plt.title("data by make_circles()")
plt.xlabel("Original Data")


# K-Means 聚类结果
plt.subplot(1, 2, 2)

plt.scatter(
    x[:, 0],
    x[:, 1],
    c=y_pred
)

# 绘制聚类中心
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    marker="*",
    s=300
)

plt.title("data by make_circles()")
plt.xlabel("K-Means")

plt.show()
```

# 第五题


```python
import os
import matplotlib.pyplot as plt
import numpy as np
data_dir = "/Users/caijingxi/Desktop/机器学习/第三次作业/face_images"

person_names = []

#检查读取资料架
for name in os.listdir(data_dir):

    path = os.path.join(data_dir, name)

    if os.path.isdir(path):
        person_names.append(name)

person_names = sorted(person_names)

# print(person_names)
# print("人数：", len(person_names))


#确认读取照片尺寸
import matplotlib.pyplot as plt

person = person_names[0]

folder = os.path.join(data_dir, person)

files = sorted(os.listdir(folder))

# print(files[:5])
img_path = os.path.join(folder, files[0])

img = plt.imread(img_path)

# print(img.shape)

# plt.imshow(img)
# plt.show()

#读取全部图片
images = []
labels = []

for label, person in enumerate(person_names):

    folder = os.path.join(data_dir, person)

    for file in sorted(os.listdir(folder)):

        if file.endswith(".jpg"):

            img_path = os.path.join(folder, file)

            img = plt.imread(img_path)

            images.append(img)

            labels.append(label)
            
images = np.array(images)
y = np.array(labels)

# print("images shape:", images.shape)
# print("y shape:", y.shape)

#二维压缩成一维
X = images.reshape(len(images), -1)

#开始聚类
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=len(person_names),
    random_state=42
)

kmeans.fit(X)

y_pred = kmeans.labels_

#老师提供的函数
from sklearn import metrics
import numpy as np
#旧版
# from sklearn.utils.linear_assignment_ import linear_assignment
from scipy.optimize import linear_sum_assignment
'''
   Clustering accuracy
'''

def cluster_acc(y_true, y_pred):
    """
    Calculate clustering accuracy. Require scikit-learn installed

    # Arguments
        y: true labels, numpy.array with shape `(n_samples,)`
        y_pred: predicted labels, numpy.array with shape `(n_samples,)`

    # Return
        accuracy, in [0,1]
    """
    y_true = np.array(y_true).astype(np.int64)
    y_pred = np.array(y_pred).astype(np.int64)

    assert y_pred.size == y_true.size

    D = max(y_pred.max(), y_true.max()) + 1

    w = np.zeros((D, D), dtype=np.int64)

    for i in range(y_pred.size):
        w[y_pred[i], y_true[i]] += 1

    row_ind, col_ind = linear_sum_assignment(w.max() - w)

    return w[row_ind, col_ind].sum() * 1.0 / y_pred.size

    

'''
 Evaluation metrics of clustering performance
      ACC: clustering accuracy
      NMI: normalized mutual information
      ARI: adjusted rand index
'''
def clusteringMetrics(trueLabel, predictiveLabel):
    # Clustering accuracy
    ACC = cluster_acc(trueLabel, predictiveLabel)

    # Normalized mutual information
    NMI = metrics.normalized_mutual_info_score(trueLabel, predictiveLabel)

    # Adjusted rand index
    ARI = metrics.adjusted_rand_score(trueLabel, predictiveLabel)

    return ACC, NMI, ARI



ACC, NMI, ARI = clusteringMetrics(y, y_pred)

print("ACC =", ACC)
print("NMI =", NMI)
print("ARI =", ARI)


plt.figure(figsize=(15, 8))

for i in range(len(images)):

    plt.subplot(10, 20, i + 1)

    plt.imshow(images[i])

    plt.axis("off")

plt.tight_layout()
plt.show()
```
